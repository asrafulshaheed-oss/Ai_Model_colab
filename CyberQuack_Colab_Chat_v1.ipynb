# Cyber-Quack-14B — Colab chat v1: BF16 → Q8_0 converter • context & reasoning picker • web search (Parallel MCP) • public link

**How to use**
1. **Runtime → Change runtime type → T4 GPU** (preselected if you opened this file in Colab).
2. **One-time Hugging Face step:** `jabbatheduck/Cyber-Quack-14B` is a **gated** repo. Open the model page, sign in, click **Agree and access repository** (approval is automatic), create a READ token at `huggingface.co/settings/tokens`, and add it as the Colab secret **HF_TOKEN** (🔑 icon on the left).
3. **Cell 1** — pick the quant (default **Q8_0**) and run it. It mounts Google Drive, builds llama.cpp, downloads the **BF16 safetensors (~29.5 GB)**, converts them to F16 GGUF and quantizes **Q8_0 (~15.7 GB)**, deleting each intermediate as it goes (**≈30–50 min** the first time; re-runs skip every finished stage). On a single 16 GB T4 it also builds a **Q6_K** copy, because the Q8_0 weights (~15 GiB) physically cannot fit next to the KV cache — Cell 2 then starts on Q6_K automatically.
4. **Cell 2** — pick the **context window** (default **16,384**, the safe size on one T4), **reasoning** (default **Max**) and **web search** (default **Parallel**, free, no key), then run it. It shows the chat right in the notebook, plus a private link and a **public link**.
5. **Keep this tab open and leave Cell 2 running.** It watches the model server and the link and repairs them automatically.

**Good to know**
- **Quants:** there is **no pre-built Q8_0 GGUF** anywhere on the Hub — Cell 1 builds it here from the repo's BF16 safetensors (download → convert to F16 → quantize to Q8_0 → delete intermediates). **Q6_K** (12.1 GB, near-lossless) is the one that actually fits a single free T4 with usable context.
- **Context:** Cyber-Quack-14B is Qwen2.5-14B — **native context 32,768** (no YaRN up to there). Q8_0 14B on a 16 GB T4 leaves only ~2–3 GB for KV cache: 32k is the practical ceiling and the notebook starts at 16k and auto-shrinks on OOM. For more context, use the **Kaggle 2×T4 notebook**. Very long chats never crash the server: context shift is enabled, so when the window fills, old tokens are recycled instead of erroring.
- **Reasoning:** Cyber-Quack-14B (Qwen2.5 base) has **no native `<think>` blocks**. The Reasoning dropdown still maps onto llama-server's `--reasoning` / `--reasoning-budget` flags (Max/High/Medium/Low/Off = unlimited/8192/2048/512/0 thinking tokens); the collapsible reasoning panel in the UI fills only if the model emits think content. You can also switch it per chat in the UI.
- **Web search:** the model gets `web_search` and `web_fetch` tools (Parallel + Exa + DuckDuckGo with auto-fallback; a Parallel key adds the deep `web_research` tool — the keys are pre-filled). The first search asks permission in the chat — click **Always allow** — or use the **Allow tool calls: ON / OFF** switch inside the chat box, right after the + button. It is on by default, remembered in your browser, and applies instantly without reloading.
- **Public link:** uses a **Cloudflare** link (no account, new address each run). The chat inside the notebook always works too.
- **Long input:** 1,000+ lines of text or code is fine — paste it or attach the file.
- **No vision:** no image projector (mmproj) exists for this model, so this notebook is text-only.
- Chats are stored in your browser tab. If the public link changes, it's a new origin, so export chats you care about from the UI settings.


```python
#@title 1️⃣ Setup — build llama.cpp + convert Cyber-Quack-14B to Q8_0 (≈30–50 min first run; re-runs skip it) { display-mode: "form" }
MODEL_QUANT = "Q8_0 · 15.7 GB · best quality (default)" #@param ["Q8_0 · 15.7 GB · best quality (default)", "Q6_K · 12.1 GB · near-lossless, fits one T4 with 16k context"]
USE_DRIVE_CACHE = True #@param {type:"boolean"}
# MODEL_QUANT: there is NO pre-built Q8_0 GGUF on the Hub — this cell downloads the BF16
#   safetensors (~29.5 GB) and runs the conversion + Q8_0 quantization right here.
#   On a single free T4 the Q8_0 copy cannot load (weights ~15 GiB vs 15 GB VRAM), so on that
#   hardware the cell also builds a Q6_K copy and Cell 2 falls back to it automatically.
# USE_DRIVE_CACHE: mounts Google Drive and caches the compiled llama.cpp binaries there, so
#   the next session skips the ~10 min build.
# ⚠️ GATED REPO: jabbatheduck/Cyber-Quack-14B requires a Hugging Face token that has accepted
#   its terms. Accept once on the model page, then add a READ token as the Colab secret
#   HF_TOKEN (🔑 icon on the left). Details in the markdown above.

import os, sys, time, json, shutil, subprocess
from concurrent.futures import ThreadPoolExecutor

SCRATCH = "/content"

# ---------- Google Drive first (optional cache; must mount on the main thread) ----------
if USE_DRIVE_CACHE:
    try:
        from google.colab import drive
        drive.mount("/content/drive")
        print("✅ Google Drive mounted (build cache)")
    except Exception as e:
        USE_DRIVE_CACHE = False
        print(f"⚠️ Could not mount Google Drive ({e}); continuing without the build cache.")

QUANT      = MODEL_QUANT.split()[0]                     # "Q8_0" or "Q6_K"
REPO_ID    = "jabbatheduck/Cyber-Quack-14B"             # BF16 safetensors only (no pre-built GGUF)
LLAMA_TAG  = "b11925"                                   # pinned llama.cpp release (tested), so a new upstream change can't break this notebook

LLAMA_DIR  = "/content/llama.cpp"
BIN_PATH   = f"{LLAMA_DIR}/build/bin/llama-server"
QBIN_PATH  = f"{LLAMA_DIR}/build/bin/llama-quantize"
STAMP      = f"{LLAMA_DIR}/build/cyberquack_build_stamp.txt"
MODEL_DIR  = f"{SCRATCH}/models"
BF16_DIR   = f"{MODEL_DIR}/cyber-quack-bf16"
F16_GGUF   = f"{SCRATCH}/cyber-quack-f16.gguf"
FINAL_GGUF = f"{SCRATCH}/cyber-quack-{QUANT.lower()}.gguf"
FALLBACK_GGUF = f"{SCRATCH}/cyber-quack-q6_k.gguf"
CF_PATH    = "/content/cloudflared"
CONFIG     = "/content/cyberquack_config.json"
BUILD_LOG  = "/content/build.log"
CF_LOG     = "/content/cf.log"
CONVERT_LOG = "/content/convert.log"
QUANT_LOG  = "/content/quantize.log"
UI_TGZ     = "/content/llama-ui.tar.gz"
CACHE_DIR  = "/content/drive/MyDrive/cyber_quack_cache"
os.environ["PATH"] += ":/usr/local/cuda/bin"
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.makedirs(MODEL_DIR, exist_ok=True)
USE_CACHE = USE_DRIVE_CACHE

# ---------- GPU check ----------
try:
    gpu = subprocess.check_output(["nvidia-smi", "--query-gpu=name,compute_cap,memory.total",
                                   "--format=csv,noheader"], text=True).strip().splitlines()[0]
    gpu_name, cap, vram = [x.strip() for x in gpu.split(",")]
    ARCH = cap.replace(".", "")
    print(f"✅ GPU: {gpu_name} (sm_{ARCH}, {vram})")
except Exception:
    raise RuntimeError("❌ No GPU found. Runtime → Change runtime type → T4 GPU → Save, then run this cell again.")

# ---------- disk guard (the BF16 → F16 → Q8_0 pipeline needs real scratch space) ----------
free_gb = shutil.disk_usage(SCRATCH).free / 1e9
if free_gb < 35:
    raise RuntimeError(
        f"❌ Only {free_gb:.0f} GB free in {SCRATCH}. The Cyber-Quack pipeline peaks at ~60 GB "
        "(BF16 download + F16 GGUF exist at the same time during conversion). "
        "Runtime → Disconnect and delete runtime → reconnect to get a fresh disk, then run again. "
        "(Colab T4 runtimes normally have ~80 GB — a nearly-full Google Drive disk or an old "
        "session is the usual cause.)")
if free_gb < 62:
    print(f"⚠️ {free_gb:.0f} GB free in {SCRATCH}; the conversion peaks at ~60 GB. It should still fit — "
          "live disk numbers print below. If it runs out, the cell auto-retries converting straight to Q8_0.")

if USE_DRIVE_CACHE:
    os.makedirs(CACHE_DIR, exist_ok=True)

# binary cache names need the GPU arch, which is only known after the check above
CACHED_BIN   = f"{CACHE_DIR}/llama-server-{LLAMA_TAG}-sm{ARCH}"
CACHED_QUANT = f"{CACHE_DIR}/llama-quantize-{LLAMA_TAG}-sm{ARCH}"

def run(cmd, log, env=None, retries=1):
    for attempt in range(retries):
        with open(log, "a") as f:
            f.write(f"\n$ {' '.join(cmd)}\n"); f.flush()
            p = subprocess.run(cmd, stdout=f, stderr=subprocess.STDOUT, env=env)
        if p.returncode == 0:
            return
        if attempt + 1 < retries:
            time.sleep(5)
    tail = open(log, errors="replace").read()[-4000:]
    raise RuntimeError(f"Command failed: {' '.join(cmd)}\n\n{tail}")

def binary_ok(path):
    try:
        out = subprocess.run([path, "--version"], capture_output=True, text=True, timeout=60)
        return out.returncode == 0
    except Exception:
        return False

# Builds BOTH llama-server and llama-quantize (the quantizer is required by the
# Cyber-Quack BF16 -> Q8_0 pipeline; a plain chat build only needs llama-server).
def build_server():
    want = f"{LLAMA_TAG} sm{ARCH}"
    have_server = os.path.exists(BIN_PATH) and binary_ok(BIN_PATH)
    have_quant = os.path.exists(QBIN_PATH) and binary_ok(QBIN_PATH)
    if have_server and have_quant and os.path.exists(STAMP) and open(STAMP).read().strip() == want:
        return "already built"
    if USE_CACHE:
        os.makedirs(os.path.dirname(BIN_PATH), exist_ok=True)
        if os.path.exists(CACHED_BIN):
            shutil.copy(CACHED_BIN, BIN_PATH); os.chmod(BIN_PATH, 0o755)
            have_server = binary_ok(BIN_PATH)
        if os.path.exists(CACHED_QUANT):
            shutil.copy(CACHED_QUANT, QBIN_PATH); os.chmod(QBIN_PATH, 0o755)
            have_quant = binary_ok(QBIN_PATH)
        if have_server and have_quant:
            open(STAMP, "w").write(want)
            return "restored from cache"
    log = BUILD_LOG
    open(log, "w").close()
    shutil.rmtree(LLAMA_DIR, ignore_errors=True)        # always start from the pinned, clean source
    run(["git", "clone", "--depth", "1", "--branch", LLAMA_TAG,
         "https://github.com/ggml-org/llama.cpp", LLAMA_DIR], log, retries=3)
    # Web UI: use the official prebuilt UI of this exact release (no Node.js build needed)
    dist = f"{LLAMA_DIR}/tools/ui/dist"
    os.makedirs(dist, exist_ok=True)
    try:
        run(["wget", "-q", "--tries=3", "-O", UI_TGZ,
             f"https://github.com/ggml-org/llama.cpp/releases/download/{LLAMA_TAG}/llama-{LLAMA_TAG}-ui.tar.gz"], log)
        run(["tar", "-xzf", UI_TGZ, "-C", dist, "--strip-components=1"], log)
    except Exception:
        shutil.rmtree(dist, ignore_errors=True)         # cmake then fetches the same UI version from Hugging Face
    env = dict(os.environ, HF_UI_VERSION=LLAMA_TAG)
    run(["cmake", "-S", LLAMA_DIR, "-B", f"{LLAMA_DIR}/build",
         "-DGGML_CUDA=ON", f"-DCMAKE_CUDA_ARCHITECTURES={ARCH}",
         "-DBUILD_SHARED_LIBS=OFF", "-DLLAMA_OPENSSL=OFF",
         "-DLLAMA_BUILD_TESTS=OFF", "-DLLAMA_BUILD_EXAMPLES=OFF",
         "-DCMAKE_BUILD_TYPE=Release"], log, env=env)
    run(["cmake", "--build", f"{LLAMA_DIR}/build", "--config", "Release",
         "--target", "llama-server", "--target", "llama-quantize", "-j", str(os.cpu_count() or 2)], log, env=env)
    if not binary_ok(BIN_PATH):
        raise RuntimeError(f"The server was built but does not start. See {BUILD_LOG}")
    if not binary_ok(QBIN_PATH):
        raise RuntimeError(f"llama-quantize was not built. See {BUILD_LOG}")
    open(STAMP, "w").write(want)
    if USE_CACHE:
        shutil.copy(BIN_PATH, CACHED_BIN)
        shutil.copy(QBIN_PATH, CACHED_QUANT)
    return "built from source"

def get_convert_deps():
    # everything convert_hf_to_gguf.py imports (torch/transformers/gguf/safetensors/...),
    # plus the DuckDuckGo web-search deps - installed in ONE pip process: two pip runs at the
    # same time can corrupt each other's metadata.
    p = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--no-input", "-U",
                        "gguf", "ddgs", "trafilatura"], capture_output=True, text=True)
    pip_ok = p.returncode == 0
    need = []
    for mod, pkg in (("torch", "torch"), ("transformers", "transformers"), ("safetensors", "safetensors"),
                     ("sentencepiece", "sentencepiece"), ("numpy", "numpy"),
                     ("google.protobuf", "protobuf")):
        try:
            __import__(mod)
        except ImportError:
            need.append(pkg)
    if need:
        cmd = [sys.executable, "-m", "pip", "install", "-q", "--no-input"] + need
        if "torch" in need:
            # CPU wheel is fine: the conversion is CPU-bound (keeps the CUDA torch untouched when present)
            cmd += ["--extra-index-url", "https://download.pytorch.org/whl/cpu"]
        p2 = subprocess.run(cmd, capture_output=True, text=True)
        if p2.returncode != 0:
            return f"pip failed ({(p2.stderr or p2.stdout).strip()[-160:]})"
    return "installed" if pip_ok else "installed (preinstalled torch/transformers kept)"


import threading

class Heartbeat(threading.Thread):
    """Prints a progress line every few seconds while a download/conversion runs."""
    def __init__(self, label, watch, is_dir, interval=10, total_bytes=0):
        super().__init__(daemon=True)
        self.label, self.watch, self.is_dir = label, watch, is_dir
        self.interval, self.total = interval, total_bytes
        self.t0 = time.time()
        self.stop_event = threading.Event()

    def _size(self):
        try:
            if self.is_dir:
                total = 0
                for root, _, files in os.walk(self.watch):
                    for name in files:
                        try:
                            total += os.path.getsize(os.path.join(root, name))
                        except OSError:
                            pass
                return total
            return os.path.getsize(self.watch) if os.path.exists(self.watch) else 0
        except OSError:
            return 0

    def run(self):
        while not self.stop_event.wait(self.interval):
            cur = self._size()
            msg = f"   [{self.label}] {int((time.time() - self.t0) // 60)}:{int((time.time() - self.t0) % 60):02d} | {cur/1e9:5.1f} GB"
            if self.total:
                msg += f" / ~{self.total/1e9:.0f} GB"
            try:
                msg += f" | {shutil.disk_usage(SCRATCH).free/1e9:5.1f} GB free"
            except OSError:
                pass
            print(msg, flush=True)

    def stop(self):
        self.stop_event.set()
        self.join(timeout=5)


def last_log_line(log):
    try:
        lines = [l for l in open(log, errors="replace").read().splitlines() if l.strip()]
        return lines[-1][:110] if lines else ""
    except OSError:
        return ""

def run_with_progress(cmd, log, label, watch_path, is_dir, interval):
    """Run a subprocess while printing progress (file/dir size + free disk) every `interval` s."""
    with open(log, "a") as f:
        f.write(f"\n$ {' '.join(cmd)}\n"); f.flush()
        proc = subprocess.Popen(cmd, stdout=f, stderr=subprocess.STDOUT)
    hb = Heartbeat(label, watch_path, is_dir, interval)
    hb.start()
    try:
        while proc.poll() is None:
            time.sleep(1)
    finally:
        hb.stop()
    return proc.returncode

def hf_token():
    try:
        from google.colab import userdata
        return (userdata.get("HF_TOKEN") or "").strip()
    except Exception:
        return ""

GATED_HELP = (
    "❌ jabbatheduck/Cyber-Quack-14B is a GATED Hugging Face repo: downloads need a token that has accepted its terms.\n"
    "   1. Open https://huggingface.co/jabbatheduck/Cyber-Quack-14B and sign in.\n"
    "   2. Click “Agree and access repository” (approval is automatic).\n"
    "   3. Create a READ token at https://huggingface.co/settings/tokens.\n"
    "   4. Add it to this notebook as a secret named HF_TOKEN (Colab: the 🔑 icon on the left).\n"
    "   5. Run this cell again."
)

def download_bf16():
    """~29.5 GB of BF16 safetensors into BF16_DIR (resumable; skips completed files on re-run)."""
    try:
        import huggingface_hub  # noqa
    except ImportError:
        subprocess.run([sys.executable, "-m", "pip", "-q", "install", "huggingface_hub"], check=True)
    from huggingface_hub import snapshot_download, HfApi
    token = hf_token()
    total = 0
    try:
        info = HfApi(token=token or None).model_info(REPO_ID, files_metadata=True)
        total = sum(s.size or 0 for s in info.siblings if (s.rfilename or "").endswith(".safetensors"))
    except Exception:
        total = int(29.5e9)   # estimate for Qwen2.5-14B BF16
    ignore = ["*.gguf", "*.md", "*.png", "*.jpg"]   # the repo has no GGUFs; this keeps any junk out
    last = None
    for attempt in range(4):
        hb = Heartbeat("download", BF16_DIR, True, 10, total)
        hb.start()
        try:
            snapshot_download(REPO_ID, local_dir=BF16_DIR, ignore_patterns=ignore, token=token or None)
            return
        except Exception as e:
            last = e
            msg = str(e)
            if "401" in msg or "403" in msg or "restricted" in msg.lower() or "gated" in msg.lower() or "log in" in msg.lower():
                raise RuntimeError(GATED_HELP) from e
            print(f"⚠️ download attempt {attempt + 1}/4 failed: {msg[:200]}")
            time.sleep(10 * (attempt + 1))
        finally:
            hb.stop()
    raise RuntimeError(f"BF16 download failed after 4 tries: {last}")

def convert_stage():
    """BF16 safetensors -> F16 GGUF. Returns "f16" or "direct-q8" (low-disk rescue)."""
    print("⏳ Converting BF16 → F16 GGUF (CPU-bound, ~10–25 min)...")
    free0 = shutil.disk_usage(SCRATCH).free / 1e9
    print(f"   free disk before: {free0:.1f} GB")
    log = CONVERT_LOG
    open(log, "w").close()
    cmd = [sys.executable, f"{LLAMA_DIR}/convert_hf_to_gguf.py", BF16_DIR,
           "--outfile", F16_GGUF, "--outtype", "f16"]
    rc = run_with_progress(cmd, log, "convert", F16_GGUF, False, 10)
    free1 = shutil.disk_usage(SCRATCH).free / 1e9
    print(f"   free disk after: {free1:.1f} GB (before: {free0:.1f} GB)")
    if rc != 0:
        tail = open(log, errors="replace").read()[-2500:]
        if "no space left" in tail.lower() or "errno 28" in tail.lower() or "os error 28" in tail.lower():
            if os.path.exists(F16_GGUF):
                os.remove(F16_GGUF)
            print("⚠️ Disk ran out during the F16 conversion — converting straight to Q8_0 instead "
                  "(needs ~30 GB less peak scratch; the result is the same Q8_0 quant).")
            cmd = [sys.executable, f"{LLAMA_DIR}/convert_hf_to_gguf.py", BF16_DIR,
                   "--outfile", f"{SCRATCH}/cyber-quack-q8_0.gguf", "--outtype", "q8_0"]
            if run_with_progress(cmd, log, "convert-q8", f"{SCRATCH}/cyber-quack-q8_0.gguf", False, 10) == 0:
                return "direct-q8"
            tail = open(log, errors="replace").read()[-2500:]
        raise RuntimeError(f"BF16 → GGUF conversion failed:\n{tail}")
    return "f16"

def ensure_quantize_tool():
    """llama-quantize comes from the same build; build just this target if it is missing
    (e.g. the server was restored from cache but the scratch build dir is gone)."""
    if os.path.exists(QBIN_PATH) and binary_ok(QBIN_PATH):
        return
    os.makedirs(os.path.dirname(QBIN_PATH), exist_ok=True)
    log = BUILD_LOG
    if not os.path.exists(f"{LLAMA_DIR}/CMakeLists.txt"):
        shutil.rmtree(LLAMA_DIR, ignore_errors=True)
        run(["git", "clone", "--depth", "1", "--branch", LLAMA_TAG,
             "https://github.com/ggml-org/llama.cpp", LLAMA_DIR], log, retries=3)
    env = dict(os.environ, HF_UI_VERSION=LLAMA_TAG)
    run(["cmake", "-S", LLAMA_DIR, "-B", f"{LLAMA_DIR}/build",
         "-DGGML_CUDA=ON", f"-DCMAKE_CUDA_ARCHITECTURES={ARCH}",
         "-DBUILD_SHARED_LIBS=OFF", "-DLLAMA_OPENSSL=OFF",
         "-DLLAMA_BUILD_TESTS=OFF", "-DLLAMA_BUILD_EXAMPLES=OFF",
         "-DCMAKE_BUILD_TYPE=Release"], log, env=env)
    run(["cmake", "--build", f"{LLAMA_DIR}/build", "--config", "Release",
         "--target", "llama-quantize", "-j", str(os.cpu_count() or 2)], log, env=env)
    if not binary_ok(QBIN_PATH):
        raise RuntimeError(f"llama-quantize build failed. See {BUILD_LOG}")

def quantize_stage():
    """F16 GGUF -> Q8_0 GGUF (the F16 file is deleted by prep_model afterwards)."""
    ensure_quantize_tool()
    print(f"⏳ Quantizing F16 → {QUANT} (CPU-bound, ~8–15 min)...")
    log = QUANT_LOG
    open(log, "w").close()
    cmd = [QBIN_PATH, F16_GGUF, FINAL_GGUF, QUANT]
    rc = run_with_progress(cmd, log, "quantize", FINAL_GGUF, False, 5)
    if rc != 0 or not os.path.exists(FINAL_GGUF):
        raise RuntimeError(f"Quantization failed (exit {rc}):\n" + open(log, errors="replace").read()[-2500:])
    print(f"✅ Quantized: {os.path.basename(FINAL_GGUF)} ({os.path.getsize(FINAL_GGUF)/1e9:.1f} GB)")

def maybe_quantize_fallback():
    """On a single 16 GB GPU the Q8_0 copy physically cannot load (weights alone ~15 GiB).
    Also quantize a Q6_K copy from the same F16 so Cell 2 can fall back to it."""
    if QUANT != "Q8_0" or not os.path.exists(F16_GGUF):
        return ""
    vram = total_vram_mib()
    if not vram or vram >= 17 * 1024:
        return ""      # 2×T4 / bigger GPUs hold Q8_0 fine
    if shutil.disk_usage(SCRATCH).free / 1e9 < 14:
        print("⚠️ Less than 14 GB of disk left — skipping the Q6_K fallback copy.")
        return ""
    print("⏳ Single ~16 GB GPU detected: also building a Q6_K copy (12.1 GB) that Cell 2 "
          "falls back to if Q8_0 can't fit...")
    ensure_quantize_tool()
    log = QUANT_LOG
    cmd = [QBIN_PATH, F16_GGUF, FALLBACK_GGUF, "Q6_K"]
    rc = run_with_progress(cmd, log, "quantize-q6k", FALLBACK_GGUF, False, 5)
    if rc != 0 or not os.path.exists(FALLBACK_GGUF):
        print("⚠️ Q6_K fallback quantization failed (continuing with Q8_0 only); see the log.")
        return ""
    return FALLBACK_GGUF

def total_vram_mib():
    try:
        out = subprocess.check_output(["nvidia-smi", "--query-gpu=memory.total",
                                       "--format=csv,noheader,nounits"], text=True, stderr=subprocess.DEVNULL)
        return sum(int(x.strip()) for x in out.strip().splitlines())
    except Exception:
        return 0

def cleanup_old_quants(keep_names):
    """Delete cyber-quack GGUFs other than the ones we keep (no orphaned files)."""
    for name in os.listdir(SCRATCH):
        if (name.startswith("cyber-quack-") and name.endswith(".gguf")
                and name not in keep_names):
            try:
                os.remove(os.path.join(SCRATCH, name))
                print(f"   removed old {name}")
            except OSError:
                pass

def prep_model():
    """The full BF16 → F16 → Q8_0 → cleanup chain (staged: re-runs resume where they stopped).
    Returns (model_path, fallback_path, quant_label)."""
    t0 = time.time()
    # 0) fast path: the final quant already exists from an earlier run (>10 GB = complete)
    if os.path.exists(FINAL_GGUF) and os.path.getsize(FINAL_GGUF) > 10_000_000_000:
        fallback = ""
        if QUANT == "Q8_0" and os.path.exists(FALLBACK_GGUF) and os.path.getsize(FALLBACK_GGUF) > 10_000_000_000:
            fallback = FALLBACK_GGUF
        print(f"✅ {os.path.basename(FINAL_GGUF)} already exists "
              f"({os.path.getsize(FINAL_GGUF)/1e9:.1f} GB) — skipping download/conversion/quantization.")
        keep = {os.path.basename(FINAL_GGUF)}
        if fallback:
            keep.add(os.path.basename(fallback))
        shutil.rmtree(BF16_DIR, ignore_errors=True)   # stale intermediates from crashed runs
        if os.path.exists(F16_GGUF):
            os.remove(F16_GGUF)
        cleanup_old_quants(keep)
        return FINAL_GGUF, fallback, QUANT
    # 1) conversion deps (installed while llama.cpp compiles in the other thread)
    print("✅ Conversion deps: " + str(get_convert_deps()))
    # 2) BF16 source -> F16 GGUF
    if os.path.exists(F16_GGUF) and os.path.getsize(F16_GGUF) > 20_000_000_000:
        print("✅ F16 GGUF already converted — resuming at quantization.")
        how = "f16"
    else:
        print(f"⏳ Downloading the BF16 safetensors of {REPO_ID} (~29.5 GB)...")
        download_bf16()
        how = convert_stage()
    # 3) delete the BF16 safetensors dir immediately after conversion (~29.5 GB freed)
    shutil.rmtree(BF16_DIR, ignore_errors=True)
    print("✅ Deleted the BF16 safetensors dir (~29.5 GB freed).")
    # 4) F16 -> chosen quant (+ Q6_K fallback copy on a single-16 GB-GPU machine), then cleanup
    served, fallback, label = FINAL_GGUF, "", QUANT
    if how == "f16":
        quantize_stage()
        fallback = maybe_quantize_fallback() or ""
    else:
        # low-disk rescue: the Q8_0 file was written directly from BF16
        served = f"{SCRATCH}/cyber-quack-q8_0.gguf"
        label = "Q8_0"
        if QUANT != "Q8_0":
            print(f"ℹ️ Serving Q8_0 (direct-conversion rescue) instead of {QUANT}: there is no F16 file to quantize from.")
    if os.path.exists(F16_GGUF):
        os.remove(F16_GGUF)
        print("✅ Deleted the F16 GGUF — only the quantized file(s) are kept.")
    # 5) no orphaned files: keep just the final copy (and the Q6_K fallback, if present)
    keep = {os.path.basename(served)}
    if fallback:
        keep.add(os.path.basename(fallback))
    cleanup_old_quants(keep)
    print(f"✅ Pipeline done in {int((time.time()-t0)//60)} min {int((time.time()-t0)%60):02d} s | "
          f"{shutil.disk_usage(SCRATCH).free/1e9:.1f} GB free.")
    return served, fallback, label

def get_cloudflared():
    if os.path.exists(CF_PATH) and os.path.getsize(CF_PATH) > 1_000_000 and binary_ok(CF_PATH):
        return "already there"
    run(["wget", "-q", "--tries=3", "-O", CF_PATH,
         "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64"],
        CF_LOG, retries=3)
    os.chmod(CF_PATH, 0o755)
    if os.path.getsize(CF_PATH) < 1_000_000 or not binary_ok(CF_PATH):
        os.remove(CF_PATH)
        raise RuntimeError("cloudflared download looks broken; run this cell again.")
    return "downloaded"

t0 = time.time()
print("⏳ " + str(f"Building llama.cpp {LLAMA_TAG} and preparing {REPO_ID} ({QUANT} from BF16) at the same time..."))
with ThreadPoolExecutor(max_workers=3) as ex:
    jobs = {"build": ex.submit(build_server), "model": ex.submit(prep_model),
           "tunnel tool": ex.submit(get_cloudflared)}
    last = -1
    while not all(j.done() for j in jobs.values()):
        time.sleep(5)
        mins = int((time.time() - t0) // 60)
        if mins != last:
            last = mins
            print(f"   {mins:>2} min | " + " | ".join(f"{k}: {'done' if j.done() else '…'}" for k, j in jobs.items()))
    errors = []
    for k, j in jobs.items():
        if j.exception() is not None and k != "__none__":
            errors.append(f"[{k}] {j.exception()}")
    if errors:
        raise RuntimeError("Setup failed — just run this cell again (finished parts are reused).\n\n" + "\n\n".join(errors))
build_status = jobs["build"].result()
model_path, fallback_path, quant_label = jobs["model"].result()
cf_status = jobs["tunnel tool"].result()

cfg = {"model": model_path, "fallback_model": fallback_path, "quant": quant_label, "bin": BIN_PATH,
       "cloudflared": CF_PATH, "tag": LLAMA_TAG, "ui_tag": LLAMA_TAG}
json.dump(cfg, open(CONFIG, "w"))

print(f"\n✅ Server:      {build_status} (llama.cpp {LLAMA_TAG})")
print(f"✅ Model:       {os.path.basename(model_path)} ({os.path.getsize(model_path)/1e9:.1f} GB)")
if fallback_path:
    print(f"✅ Fallback:    {os.path.basename(fallback_path)} (Cell 2 uses it automatically if Q8_0 can't fit)")
print(f"✅ Tunnel tool: {cf_status}")
print(f"⏱  {int(time.time() - t0)//60} min {int(time.time() - t0)%60} s.  👉 Now run Cell 2.")

```

```python
#@title 2️⃣ Launch — choose context, reasoning and web search, then run (keep this cell running!) { display-mode: "form" }
CONTEXT_WINDOW = "16384 · safe default for one T4 (default)" #@param ["32768 · native max (model limit)", "24576", "16384 · safe default for one T4 (default)", "8192", "4096"]
REASONING = "Max · unlimited thinking (default)" #@param ["Max · unlimited thinking (default)", "High · up to 8192 thinking tokens", "Medium · up to 2048 thinking tokens", "Low · up to 512 thinking tokens", "Off · answers directly, no thinking"]
WEB_SEARCH = "Parallel · free, no key needed (default)" #@param ["Parallel · free, no key needed (default)", "Exa · free tier is rate-limited, best with a key", "DuckDuckGo · free, no key needed", "Off · no web tools"]
AUTO_FALLBACK = True #@param {type:"boolean"}
PARALLEL_API_KEY = "VtYAbqEv6_sOvxUtcGeHI28jHmIpGuxM_talrLr6" #@param {type:"string"}
EXA_API_KEY = "29334ebe-08ca-4c98-8b57-af84986d1464" #@param {type:"string"}
PUBLIC_LINK = "Cloudflare · no account, new link each run (default)" #@param ["Cloudflare · no account, new link each run (default)", "Off · private Colab link only"]
SHOW_INLINE = True #@param {type:"boolean"}
# CONTEXT_WINDOW: if your choice doesn't fit in GPU memory, the next smaller size is used automatically (you'll be told).
#   32,768 is Cyber-Quack's native limit (Qwen2.5). Bigger sizes need YaRN — use the Kaggle 2×T4 notebook for those.
# REASONING: Cyber-Quack-14B (Qwen2.5) has no native <think> blocks; the dropdown maps onto llama-server's
#   --reasoning/--reasoning-budget flags, and the UI's reasoning panel fills only if the model emits think content.
# AUTO_FALLBACK: if the chosen search service is down or rate-limited, the other free ones are tried automatically.
# API keys are hardcoded above (source of truth). Colab secrets named PARALLEL_API_KEY / EXA_API_KEY are only
#   used if the field above is emptied. A Parallel key raises limits and adds a deep "research" tool.
# PUBLIC_LINK: anyone with the link can use your session. Cloudflare needs no account; the address changes each run.
# SHOW_INLINE: also shows the chat right here in the notebook (private, always works).
# ---------------- researched defaults for Cyber-Quack-14B on one Colab T4 ----------------
# Model card (jabbatheduck/Cyber-Quack-14B = Qwen2.5-14B): 48 layers, 8 KV heads, head_dim 128,
# native context 32,768 (max_position_embeddings=32768; 131,072 possible with YaRN).
# KV cache at q8_0 ≈ 102 KiB/token. Free T4 VRAM: 15,095 MiB.
# Q8_0 weights ≈ 14,990 MiB → they cannot fit on ONE T4 at any context, so Cell 1 also builds a
# Q6_K copy (≈11,580 MiB) and this cell falls back to it automatically. With Q6_K:
#   16,384 ctx ≈ 1,632 MiB KV → ~1.4 GiB free ✓ (default) · 24,576 → ~0.6 GiB ✓ · 32,768 → ✗.
# Default context = 16,384 · dropdown max = 32,768 (native; no YaRN) · reasoning = Max · unlimited.

# ---------------- advanced (rarely needed) ----------------
KV_CACHE_TYPE    = "q8_0"      # "q8_0" halves context memory vs "f16" with almost no quality loss
TEMP, TOP_P, TOP_K, MIN_P = 1.0, 0.95, 20, 0.0   # sampling defaults
PRESENCE_PENALTY = 0.0         # if answers ever loop/repeat, set 1.0
MAX_SERVER_RESTARTS = 5
# -----------------------------------------------------------
import os, re, sys, json, time, datetime, subprocess, shutil
import requests
from IPython.display import display, HTML

import socket
def free_port(start):
    """First port from `start` that nothing else is using."""
    for p in range(start, start + 100):
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sk:
            try:
                sk.bind(("127.0.0.1", p))
                return p
            except OSError:
                continue
    raise RuntimeError(f"No free port found near {start}")

LOG, TLOG = "/content/llama-server.log", "/content/tunnel.log"
CONFIG = "/content/cyberquack_config.json"
WEB_SCRIPT, MCP_CONFIG = "/content/cyberquack_web_mcp.py", "/content/cyberquack_mcp.json"
CTX_STEPS = [32768, 24576, 16384, 8192, 4096]
NATIVE_CTX = 32768          # model's native context (above this = YaRN territory)
MIN_FREE_MIB = 400            # GPU memory that must stay free after loading, so long chats don't crash later
LOCAL = requests.Session()
LOCAL.trust_env = False       # talk to the local server/tunnel directly, never through a proxy

if not os.path.exists(CONFIG):
    raise RuntimeError("Run Cell 1 first.")
cfg = json.load(open(CONFIG))
BIN, MODEL_PATH, CF, QUANT = cfg["bin"], cfg["model"], cfg["cloudflared"], cfg["quant"]
for p in (BIN, MODEL_PATH):
    if not os.path.exists(p):
        raise RuntimeError(f"Missing {p}. Run Cell 1 again.")
FALLBACK_MODEL = cfg.get("fallback_model") or ""   # Q6_K copy built by Cell 1 on single-16-GB-GPU machines

LINK_MODE = PUBLIC_LINK.split()[0].lower()            # cloudflare | off
if LINK_MODE not in ("cloudflare", "off"):
    LINK_MODE = "cloudflare"
if LINK_MODE == "cloudflare" and not os.path.exists(CF):
    raise RuntimeError("cloudflared is missing. Run Cell 1 again (or set PUBLIC_LINK to Off).")
# ---------------- parse the form ----------------
TARGET_CONTEXT = int(re.match(r"\d+", CONTEXT_WINDOW.strip()).group())
_r = REASONING.split()[0].lower()
THINKING = _r != "off"
REASONING_BUDGET = {"max": -1, "high": 8192, "medium": 2048, "low": 512, "off": 0}.get(_r, -1)
WEB_BACKEND = {"parallel": "parallel", "exa": "exa", "duckduckgo": "duckduckgo"}.get(WEB_SEARCH.split()[0].lower(), "off")

def secret(name, typed):
    if typed.strip():
        return typed.strip()
    try:
        from google.colab import userdata
        return (userdata.get(name) or "").strip()
    except Exception:
        return ""
TOGGLE_JS = r'''/* Injected by the notebook (tpuchat-v2): "Allow tool calls" switch inside the chat box,
   right after the + button. Applies instantly - never reloads the page, so a running
   answer (text or reasoning) is never interrupted. The choice is saved per browser. */
(function () {
  "use strict";
  if (window.__tpuchatAllowTools) return;
  window.__tpuchatAllowTools = true;

  var STATE_KEY = "tpuchat.allowTools";
  var TOOL_NAMES = ["web_search", "web_fetch", "web_research"];
  /* localStorage lists the official UI reads exactly once, at startup */
  var STORE_KEYS = ["LlamaUi.alwaysAllowedTools", "LlamaCppWebui.alwaysAllowedTools"];
  /* permission keys the UI derives for these tools (toolKey(): server:/mcp-<id>:/custom:) */
  var PERM_KEYS = [];
  var JUNK_KEYS = [];
  TOOL_NAMES.forEach(function (n) {
    PERM_KEYS.push("server:" + n, "mcp-web:" + n, "mcp-web:" + n.replace(/^web_/, ""), "custom:" + n);
  });
  ["web_web_search", "web_web_fetch", "web_web_research"].forEach(function (n) {
    JUNK_KEYS.push("server:" + n, "mcp-web:" + n, "custom:" + n);
  });
  TOOL_NAMES.forEach(function (n) {
    JUNK_KEYS.push("browser:" + n, "mcp:" + n);
  });

  function isOn() {
    try { var v = localStorage.getItem(STATE_KEY); return v === null ? true : v === "1"; }
    catch (e) { return true; }
  }
  function readList(k) {
    try {
      var a = JSON.parse(localStorage.getItem(k) || "[]");
      return Array.isArray(a) ? a.filter(function (x) { return typeof x === "string"; }) : [];
    } catch (e) { return []; }
  }
  function writeList(k, list) {
    try { localStorage.setItem(k, JSON.stringify(list)); } catch (e) {}
  }
  function cleanAndWrite(k, set) {
    JUNK_KEYS.forEach(function (p) { delete set[p]; });
    writeList(k, Object.keys(set));
  }
  function syncPerms(on) {
    STORE_KEYS.forEach(function (k) {
      var set = {};
      readList(k).forEach(function (x) { set[x] = 1; });
      PERM_KEYS.forEach(function (p) { if (on) set[p] = 1; else delete set[p]; });
      cleanAndWrite(k, set);
    });
  }
  function allowSaved(tool) {
    var keys = ["server:" + tool, "mcp-web:" + tool, "mcp-web:" + tool.replace(/^web_/, ""), "custom:" + tool];
    STORE_KEYS.forEach(function (k) {
      var set = {};
      readList(k).forEach(function (x) { set[x] = 1; });
      keys.forEach(function (p) { set[p] = 1; });
      cleanAndWrite(k, set);
    });
  }
  /* runs before the UI boots: the app picks this list up when it initializes */
  syncPerms(isOn());

  /* -------------- live plumbing (no reload: streaming is never interrupted) -------------- */

  var sessionAll = false;            /* "allow this session" covers all web tools until OFF again */
  var seenCard = { tool: "", t: 0 }; /* built-in permission card the user may act on */

  function isWebTool(t) { return typeof t === "string" && /^web_/.test(t); }

  /* Fetch gate - active only while the switch is OFF. Covers the one case localStorage
     cannot: web tools the UI still holds in its in-memory allowed set from an earlier
     boot, which would otherwise run silently instead of asking. */
  var origFetch = window.fetch.bind(window);
  window.fetch = function () {
    try {
      var input = arguments[0], init = arguments[1];
      var url = typeof input === "string" ? input : (input && input.url) || "";
      var method = ((init && init.method) || (input && input.method) || "GET").toUpperCase();
      if (method === "POST" && /\/tools\/?$/.test(String(url).split("?")[0]) &&
          init && typeof init.body === "string") {
        var tool = null;
        try { tool = (JSON.parse(init.body) || {}).tool; } catch (e) {}
        if (isWebTool(tool) && !isOn() && !sessionAll &&
            !(seenCard.tool === tool && Date.now() - seenCard.t < 8000)) {
          return gate(tool, arguments);
        }
      }
    } catch (e) { /* never break the app */ }
    return origFetch.apply(null, arguments);
  };

  var activeFinish = null;   /* decision callback of the currently open prompt */
  var gateChain = Promise.resolve();
  function gate(tool, args) {
    var run = function () {
      return new Promise(function (resolve) {
        var settled = false;
        var signal = args[1] && args[1].signal;
        function finish(allow, remember) {
          if (settled) return;
          settled = true;
          activeFinish = null;
          hideGateCard();
          if (remember) sessionAll = true;
          if (remember === "saved") allowSaved(tool);
          if (allow) resolve(origFetch.apply(null, args));
          else resolve(new Response(JSON.stringify({
                   error: "Blocked by the user: the Allow tool calls switch is OFF in the chat box. Turn it ON (green) to run web tools without asking."
                 }), { status: 403, statusText: "Blocked",
                      headers: { "Content-Type": "application/json" } }));
        }
        function abortNow() {
          if (settled) return;
          settled = true;
          activeFinish = null;
          hideGateCard();
          resolve(Promise.reject(new DOMException("The tool call was aborted.", "AbortError")));
        }
        if (signal) {
          if (signal.aborted) { abortNow(); return; }
          signal.addEventListener("abort", abortNow, { once: true });
        }
        showGateCard(tool, finish);
      });
    };
    gateChain = gateChain.then(run, run);
    return gateChain;
  }

  function esc(s) {
    return String(s).replace(/[&<>"']/g, function (c) {
      return { "&": "&amp;", "<": "&lt;", ">": "&gt;", '"': "&quot;", "'": "&#39;" }[c];
    });
  }

  var gateEl = null;
  function showGateCard(tool, finish) {
    hideGateCard();
    var d = document.createElement("div");
    d.id = "tpu-gate-card";
    d.setAttribute("role", "dialog");
    d.setAttribute("aria-label", "Tool call permission");
    d.innerHTML = '<div class="t">Cyber-Quack wants to run <b>' + esc(tool) + "</b></div>" +
      '<div class="s">The "Allow tool calls" switch is OFF, so this call needs your OK.</div>' +
      '<div class="row">' +
      '<button type="button" data-a="once">Allow once</button>' +
      '<button type="button" data-a="session">Allow this session</button>' +
      '<button type="button" data-a="saved">Always allow</button>' +
      '<button type="button" data-a="deny" class="deny">Deny</button>' +
      "</div>";
    d.addEventListener("click", function (e) {
      var b = e.target && e.target.closest ? e.target.closest("button[data-a]") : null;
      if (!b) return;
      var a = b.getAttribute("data-a");
      finish(a !== "deny", a === "session" ? "session" : (a === "saved" ? "saved" : null));
    });
    (document.body || document.documentElement).appendChild(d);
    gateEl = d;
    activeFinish = finish;
  }
  function hideGateCard() {
    if (gateEl) { try { gateEl.remove(); } catch (e) {} gateEl = null; }
  }
  document.addEventListener("keydown", function (e) {
    if (e.key === "Escape" && activeFinish) activeFinish(false, null);
  }, true);

  /* Watch the built-in permission card ("Allow use of <tool> ..."). While the switch is
     ON, approve it automatically (only needed when the in-memory set is missing keys,
     e.g. the switch was turned ON mid-session). While OFF, just remember it so a call
     the user just approved by hand is never asked twice. */
  function scanForCard() {
    var btns = document.getElementsByTagName("button");
    for (var i = 0; i < btns.length; i++) {
      var b = btns[i];
      if ((b.textContent || "").trim() !== "Allow once") continue;
      var root = b.parentElement, cardText = "";
      for (var k = 0; k < 7 && root; k++, root = root.parentElement) {
        cardText = root.textContent || "";
        if (cardText.indexOf("Allow use of ") >= 0) break;
      }
      if (cardText.indexOf("Allow use of ") < 0) continue;
      var m = cardText.match(/Allow use of\s+([A-Za-z0-9_.-]+)/);
      if (m) { seenCard.tool = m[1]; seenCard.t = Date.now(); }
      if (isOn() && !b.dataset.tpuDone) {
        b.dataset.tpuDone = "1";
        try { b.click(); } catch (e) {}
      }
    }
  }

  /* -------------- the switch, rendered inside the composer after the + button -------------- */

  var CSS_TEXT =
    "#tpu-allow-tools{display:inline-flex;align-items:center;gap:7px;height:32px;padding:0 12px;" +
    "border-radius:999px;border:1px solid rgba(148,163,184,.35);background:rgba(100,116,139,.14);" +
    "color:inherit;opacity:.85;font:600 12px/1 ui-sans-serif,system-ui,-apple-system,sans-serif;" +
    "cursor:pointer;user-select:none;white-space:nowrap;flex:0 0 auto;" +
    "transition:border-color .15s,background .15s,opacity .15s}" +
    "#tpu-allow-tools:hover{opacity:1;border-color:rgba(148,163,184,.6)}" +
    "#tpu-allow-tools .dot{width:8px;height:8px;border-radius:50%;background:#64748b;flex:0 0 auto;" +
    "transition:background .15s,box-shadow .15s}" +
    "#tpu-allow-tools.on{border-color:rgba(34,197,94,.55);background:rgba(34,197,94,.13);opacity:1}" +
    "#tpu-allow-tools.on .dot{background:#22c55e;box-shadow:0 0 7px rgba(34,197,94,.8)}" +
    "#tpu-allow-tools.on .lbl{color:#16a34a}" +
    "#tpu-allow-tools .lbl-short{display:none}" +
    "#tpu-gate-card{position:fixed;left:50%;bottom:96px;transform:translateX(-50%);z-index:2147483000;" +
    "width:min(440px,calc(100vw - 24px));padding:14px 16px;border-radius:14px;color:#e2e8f0;" +
    "background:#0f172a;border:1px solid rgba(148,163,184,.3);box-shadow:0 12px 40px rgba(0,0,0,.45);" +
    "font:400 13px/1.5 ui-sans-serif,system-ui,-apple-system,sans-serif;animation:tpu-pop .16s ease-out}" +
    "#tpu-gate-card .t{font-weight:600;margin-bottom:2px}" +
    "#tpu-gate-card .s{opacity:.75;margin-bottom:10px}" +
    "#tpu-gate-card .row{display:flex;flex-wrap:wrap;gap:8px}" +
    "#tpu-gate-card button{all:unset;cursor:pointer;padding:6px 12px;border-radius:8px;" +
    "font:600 12px/1 ui-sans-serif,system-ui,sans-serif;color:inherit;" +
    "border:1px solid rgba(148,163,184,.4);background:rgba(148,163,184,.12)}" +
    "#tpu-gate-card button:hover{background:rgba(148,163,184,.24)}" +
    "#tpu-gate-card button.deny{border-color:rgba(239,68,68,.5);color:#f87171}" +
    "@keyframes tpu-pop{from{opacity:0;transform:translateX(-50%) translateY(6px)}" +
    "to{opacity:1;transform:translateX(-50%) translateY(0)}}" +
    "@media (prefers-color-scheme: dark){#tpu-allow-tools.on .lbl{color:#4ade80}}" +
    "@media (prefers-color-scheme: light){#tpu-gate-card{background:#fff;color:#0f172a;" +
    "border-color:#cbd5e1;box-shadow:0 12px 40px rgba(15,23,42,.18)}}" +
    "@media (max-width:640px){#tpu-allow-tools{height:28px;padding:0 9px;gap:5px;font-size:11px}" +
    "#tpu-allow-tools .lbl-long{display:none}#tpu-allow-tools .lbl-short{display:inline}" +
    "#tpu-gate-card{bottom:84px}}";

  var chip = null;
  function paint() {
    if (!chip) return;
    var on = isOn();
    chip.classList.toggle("on", on);
    chip.setAttribute("aria-checked", on ? "true" : "false");
    var long = chip.querySelector(".lbl-long"), short = chip.querySelector(".lbl-short");
    if (long) long.textContent = "Allow tool calls: " + (on ? "ON" : "OFF");
    if (short) short.textContent = "Tools: " + (on ? "ON" : "OFF");
  }
  function buildChip() {
    try {
      var css = document.createElement("style");
      css.id = "tpu-allow-tools-css";
      css.textContent = CSS_TEXT;
      document.head.appendChild(css);
    } catch (e) {}
    var b = document.createElement("button");
    b.id = "tpu-allow-tools";
    b.type = "button";
    b.setAttribute("role", "switch");
    b.setAttribute("aria-label", "Allow tool calls");
    b.title = "ON: web tools (search / fetch) run without asking. OFF: each web tool call asks first. " +
      "Applies instantly - no reload, typing and generating continue undisturbed.";
    b.innerHTML = '<span class="dot"></span><span class="lbl lbl-long"></span><span class="lbl lbl-short"></span>';
    /* keep the composer textarea focused and never trigger the + menu or a form submit */
    ["mousedown", "pointerdown", "touchstart", "touchend"].forEach(function (ev) {
      b.addEventListener(ev, function (e) { e.preventDefault(); e.stopPropagation(); });
    });
    b.addEventListener("click", function (e) {
      e.preventDefault();
      e.stopPropagation();
      var next = !isOn();
      try { localStorage.setItem(STATE_KEY, next ? "1" : "0"); } catch (err) {}
      syncPerms(next);
      if (!next) sessionAll = false;   /* a fresh OFF period asks again */
      paint();
      /* turning ON while a call is waiting for permission: let it through */
      if (next && activeFinish) activeFinish(true, "session");
      try {
        window.dispatchEvent(new StorageEvent("storage", { key: STATE_KEY, newValue: next ? "1" : "0" }));
      } catch (err) {}
    });
    paint();
    return b;
  }
  function ensureChip() {
    var host = document.querySelector("button.file-upload-button");
    if (!host || !host.isConnected) return false;
    if (!chip) chip = buildChip();
    /* preferred spot: the left group that holds the + button, next to it */
    var group = host.closest(".mr-auto") || host.parentElement;
    if (!group) return false;
    if (chip.isConnected && chip.parentElement === group) return true;
    try { group.appendChild(chip); paint(); } catch (e) { return false; }
    return true;
  }

  window.addEventListener("storage", function (e) {
    if (e && e.key === STATE_KEY) { try { paint(); } catch (err) {} }
  });

  /* -------------- keep-alive: survive Svelte re-renders and route changes -------------- */

  var scanTimer = 0, chipTimer = 0;
  function scheduleScan() {
    if (scanTimer) return;
    scanTimer = setTimeout(function () { scanTimer = 0; try { scanForCard(); } catch (e) {} }, 120);
  }
  function scheduleChip() {
    if (chipTimer) return;
    chipTimer = setTimeout(function () { chipTimer = 0; try { ensureChip(); } catch (e) {} }, 150);
  }
  function start() {
    try {
      var mo = new MutationObserver(function (muts) {
        scheduleChip();
        for (var i = 0; i < muts.length; i++) {
          var nodes = muts[i].addedNodes;
          for (var j = 0; j < nodes.length; j++) {
            var n = nodes[j];
            if (n.nodeType === 1 && (n.textContent || "").indexOf("Allow use of ") >= 0) {
              scheduleScan();
              return;
            }
          }
        }
      });
      mo.observe(document.documentElement, { childList: true, subtree: true });
    } catch (e) {}
    setInterval(function () {
      try { ensureChip(); } catch (e) {}
      try { scanForCard(); } catch (e) {}
    }, 1500);
    try { ensureChip(); } catch (e) {}
    console.info("[tpuchat] Allow tool calls switch installed (in the chat box, next to +)");
  }
  if (document.readyState === "loading") document.addEventListener("DOMContentLoaded", start);
  else start();
})();'''


WEBUI_PATH = ""

UI_MARK = "tpuchat-v2"

def _patch_webui(index_path):
    try:
        with open(index_path, encoding="utf-8") as f:
            html = f.read()
    except OSError:
        return False
    if UI_MARK in html:
        return True
    # strip any toggle script injected by older notebook versions, then add the current one
    while True:
        stripped = re.sub(r"<script>/\* Injected by the notebook.*?</script>\s*", "", html, count=1, flags=re.S)
        if stripped == html:
            break
        html = stripped
    if "</head>" not in html:
        return False
    try:
        with open(index_path, "w", encoding="utf-8") as f:
            f.write(html.replace("</head>", "<script>" + TOGGLE_JS + "</script>\n</head>", 1))
        return True
    except OSError:
        return False

def ensure_webui():
    """Copy/patch the official web UI so it has the Allow tool calls toggle."""
    global WEBUI_PATH
    target = "/content/webui"
    index = os.path.join(target, "index.html")
    if os.path.isfile(index):
        try:
            if UI_MARK in open(index, encoding="utf-8").read():
                WEBUI_PATH = target
                return WEBUI_PATH
        except OSError:
            pass
    dist = "/content/llama.cpp/tools/ui/dist"
    try:
        if not os.path.isfile(os.path.join(dist, "index.html")):
            tag = cfg.get("ui_tag") or "b11925"
            print(f"\u23f3 Downloading the chat UI ({tag})...")
            r = requests.get(
                f"https://github.com/ggml-org/llama.cpp/releases/download/{tag}/llama-{tag}-ui.tar.gz",
                timeout=180)
            r.raise_for_status()
            tgz = "/content/webui.tar.gz"
            with open(tgz, "wb") as f:
                f.write(r.content)
            shutil.rmtree(target, ignore_errors=True)
            os.makedirs(target, exist_ok=True)
            subprocess.run(["tar", "-xzf", tgz, "-C", target, "--strip-components=1"], check=True)
            if _patch_webui(os.path.join(target, "index.html")):
                print("\u2705 Chat UI ready (with the Allow tool calls toggle)")
                WEBUI_PATH = target
                return WEBUI_PATH
        else:
            shutil.rmtree(target, ignore_errors=True)
            shutil.copytree(dist, target)
            if _patch_webui(os.path.join(target, "index.html")):
                print("\u2705 Chat UI ready (with the Allow tool calls toggle)")
                WEBUI_PATH = target
                return WEBUI_PATH
    except Exception as e:
        print(f"\u26a0\ufe0f Could not prepare the enhanced UI ({e}); using the built-in UI.")
    WEBUI_PATH = ""
    return WEBUI_PATH

PKEY = secret("PARALLEL_API_KEY", PARALLEL_API_KEY) if WEB_BACKEND != "off" else ""
EKEY = secret("EXA_API_KEY", EXA_API_KEY) if WEB_BACKEND != "off" else ""

WEB_MCP_SOURCE = r'''#!/usr/bin/env python3
"""Ornith web tools: a tiny stdio MCP server for llama-server (--mcp-servers-config).

Gives the model two tools, web_search and web_fetch, backed by:
  parallel   - Parallel MCP (https://mcp.parallel.ai/mcp), free without a key
  exa        - Exa MCP (https://mcp.exa.ai/mcp), free tier is rate limited, key optional
  duckduckgo - DuckDuckGo search (ddgs) + direct page download, no key
If the chosen backend fails (rate limit, network), it falls back to the next one
automatically, so a search almost never comes back empty.

Settings come from environment variables set by the notebook:
  WEB_BACKEND        parallel | exa | duckduckgo        (default parallel)
  PARALLEL_API_KEY   optional, higher limits + adds web_research
  EXA_API_KEY        optional
  WEB_FALLBACK       1 = try other backends on failure  (default 1)
  WEB_MAX_CHARS      cap on characters returned per call (default 24000)
Only JSON-RPC goes to stdout; diagnostics go to stderr.
"""
import json, os, re, sys, time, html, urllib.request, urllib.error

BACKEND = os.environ.get("WEB_BACKEND", "parallel").strip().lower() or "parallel"
PARALLEL_KEY = os.environ.get("PARALLEL_API_KEY", "").strip()
EXA_KEY = os.environ.get("EXA_API_KEY", "").strip()
FALLBACK = os.environ.get("WEB_FALLBACK", "1") != "0"
MAX_CHARS = int(os.environ.get("WEB_MAX_CHARS", "24000") or 24000)
UA = "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130 Safari/537.36"


def log(*a):
    print("[ornith-web]", *a, file=sys.stderr, flush=True)


# ----------------------------------------------------------------- remote MCP client
class RemoteMCP:
    """Minimal Streamable-HTTP MCP client (JSON or SSE replies)."""

    def __init__(self, url, headers=None, timeout=90):
        self.url, self.timeout, self.sid = url, timeout, None
        self.headers = {"Content-Type": "application/json",
                        "Accept": "application/json, text/event-stream",
                        "User-Agent": "ornith-colab/4"}
        self.headers.update(headers or {})
        self._id = 0

    def _post(self, payload, want_reply=True):
        h = dict(self.headers)
        if self.sid:
            h["Mcp-Session-Id"] = self.sid
        req = urllib.request.Request(self.url, data=json.dumps(payload).encode("utf-8"),
                                     headers=h, method="POST")
        try:
            with urllib.request.urlopen(req, timeout=self.timeout) as r:
                sid = r.headers.get("Mcp-Session-Id")
                if sid:
                    self.sid = sid
                ctype = (r.headers.get("Content-Type") or "").lower()
                body = r.read().decode("utf-8", errors="replace")   # always UTF-8 (fixes mojibake)
        except urllib.error.HTTPError as e:
            detail = ""
            try:
                detail = e.read().decode("utf-8", errors="replace")[:500]
            except Exception:
                pass
            raise RuntimeError(f"HTTP {e.code} from {self.url}: {detail}")
        if not want_reply:
            return None
        msgs = []
        if "event-stream" in ctype:
            data_lines = []
            for line in body.splitlines() + [""]:          # SSE: blank line ends an event
                if line.startswith("data:"):
                    data_lines.append(line[5:].lstrip(" "))
                elif line.strip() == "":
                    if data_lines:
                        try:
                            msgs.append(json.loads("\n".join(data_lines)))
                        except ValueError:
                            pass
                        data_lines = []
        elif body.strip():
            j = json.loads(body)
            msgs = j if isinstance(j, list) else [j]
        for m in msgs:
            if isinstance(m, dict) and m.get("id") == payload.get("id"):
                return m
        for m in msgs:                                      # some servers reply with id null on errors
            if isinstance(m, dict) and "error" in m:
                return m
        raise RuntimeError("No reply from MCP server")

    def _rpc(self, method, params=None):
        self._id += 1
        reply = self._post({"jsonrpc": "2.0", "id": self._id, "method": method, "params": params or {}})
        if "error" in reply:
            err = reply["error"]
            raise RuntimeError(err.get("message", str(err)) if isinstance(err, dict) else str(err))
        return reply.get("result", {})

    def call(self, tool, args):
        self._rpc("initialize", {"protocolVersion": "2025-06-18", "capabilities": {},
                                 "clientInfo": {"name": "ornith-colab", "version": "4"}})
        try:
            self._post({"jsonrpc": "2.0", "method": "notifications/initialized"}, want_reply=False)
        except Exception:
            pass
        res = self._rpc("tools/call", {"name": tool, "arguments": args})
        text = result_text(res)
        if res.get("isError"):
            raise RuntimeError(text or "tool error")
        return text


def result_text(res):
    parts = []
    for c in res.get("content") or []:
        if isinstance(c, dict) and c.get("type") == "text" and c.get("text"):
            parts.append(c["text"])
    if not parts and res.get("structuredContent") is not None:
        parts.append(json.dumps(res["structuredContent"], ensure_ascii=False, indent=1))
    return "\n".join(parts).strip()


# ----------------------------------------------------------------- backends
def _queries(a):
    q = a.get("query") or ""
    qs = a.get("search_queries") or []
    if isinstance(qs, str):
        qs = [qs]
    qs = [str(x).strip() for x in qs if str(x).strip()]
    if q.strip() and q.strip() not in qs:
        qs.insert(0, q.strip())
    return qs[:5]


def _urls(a):
    u = a.get("urls") or a.get("url") or []
    if isinstance(u, str):
        u = re.split(r"[\s,]+", u)
    out = []
    for x in u:
        x = str(x).strip().strip("<>\"'")
        if x and not re.match(r"^https?://", x, re.I):
            x = "https://" + x
        if x and x not in out:
            out.append(x)
    return out[:10]


def parallel_search(a):
    qs = _queries(a)
    objective = (a.get("objective") or "").strip() or (qs[0] if qs else "")
    hdr = {"Authorization": f"Bearer {PARALLEL_KEY}"} if PARALLEL_KEY else {}
    return RemoteMCP("https://mcp.parallel.ai/mcp", hdr).call(
        "web_search", {"objective": objective[:4000], "search_queries": qs or [objective[:200]]})


def parallel_fetch(a):
    args = {"urls": _urls(a)}
    if a.get("objective"):
        args["objective"] = str(a["objective"])[:200]
    if a.get("full_content"):
        args["full_content"] = True
    hdr = {"Authorization": f"Bearer {PARALLEL_KEY}"} if PARALLEL_KEY else {}
    return RemoteMCP("https://mcp.parallel.ai/mcp", hdr).call("web_fetch", args)


def parallel_research(a):
    hdr = {"Authorization": f"Bearer {PARALLEL_KEY}"}
    args = {"input": str(a.get("question") or a.get("input") or a.get("query") or "")[:20000]}
    return RemoteMCP("https://mcp.parallel.ai/mcp", hdr, timeout=280).call("web_research", args)


def _exa():
    hdr = {"Authorization": f"Bearer {EXA_KEY}"} if EXA_KEY else {}
    return RemoteMCP("https://mcp.exa.ai/mcp", hdr)


def exa_search(a):
    qs = _queries(a)
    args = {"query": " ".join(qs[:1]) or a.get("objective", ""), "numResults": 8}
    if a.get("objective"):
        args["objective"] = str(a["objective"])[:4000]
    return _exa().call("web_search_exa", args)


def exa_fetch(a):
    return _exa().call("web_fetch_exa", {"urls": _urls(a),
                                         "maxCharacters": 60000 if a.get("full_content") else 8000})


def ddg_search(a):
    try:
        from ddgs import DDGS
    except ImportError:
        raise RuntimeError("ddgs is not installed (pip install ddgs)")
    out, seen = [], set()
    for q in _queries(a)[:3] or [a.get("objective", "")]:
        for r in DDGS().text(q, max_results=8) or []:
            url = r.get("href") or r.get("url")
            if not url or url in seen:
                continue
            seen.add(url)
            out.append(f"## {r.get('title','')}\nURL: {url}\n{r.get('body','')}")
    if not out:
        raise RuntimeError("DuckDuckGo returned no results")
    return "\n\n".join(out)


def _download(url):
    req = urllib.request.Request(url, headers={"User-Agent": UA, "Accept": "text/html,*/*;q=0.8"})
    with urllib.request.urlopen(req, timeout=30) as r:
        raw = r.read(6_000_000)
        cs = r.headers.get_content_charset() or "utf-8"
        ctype = (r.headers.get("Content-Type") or "").lower()
    txt = raw.decode(cs, errors="replace")
    if "html" not in ctype and "<html" not in txt[:2000].lower():
        return txt
    try:
        import trafilatura
        ext = trafilatura.extract(txt, url=url, include_links=False, include_tables=True,
                                  output_format="markdown")
        if ext and len(ext) > 200:
            return ext
    except Exception:
        pass
    txt = re.sub(r"(?is)<(script|style|noscript|svg|head)[^>]*>.*?</\1>", " ", txt)
    txt = re.sub(r"(?s)<[^>]+>", " ", txt)
    return re.sub(r"[ \t\r\f\v]+", " ", re.sub(r"\n\s*\n+", "\n\n", html.unescape(txt))).strip()


def direct_fetch(a):
    urls, parts = _urls(a), []
    if not urls:
        raise RuntimeError("no valid URL given")
    per = max(3000, MAX_CHARS // len(urls))
    for u in urls:
        try:
            parts.append(f"# {u}\n{_download(u)[:per]}")
        except Exception as e:
            parts.append(f"# {u}\n(could not load: {e})")
    return "\n\n".join(parts)


SEARCH = {"parallel": parallel_search, "exa": exa_search, "duckduckgo": ddg_search}
FETCH = {"parallel": parallel_fetch, "exa": exa_fetch, "duckduckgo": direct_fetch}


def run_with_fallback(table, args):
    order = [BACKEND] + ([b for b in ("parallel", "duckduckgo", "exa") if b != BACKEND] if FALLBACK else [])
    errors = []
    for b in order:
        if b not in table:
            continue
        for attempt in range(2):                     # one quick retry per backend
            try:
                text = table[b](args)
                if text and text.strip():
                    note = "" if b == BACKEND else f"[results from {b}; {BACKEND} was unavailable]\n"
                    return note + text
                raise RuntimeError("empty result")
            except Exception as e:
                msg = str(e)
                log(f"{b} attempt {attempt + 1} failed: {msg[:300]}")
                if "rate limit" in msg.lower() or "HTTP 4" in msg:
                    break                            # don't hammer a rate-limited backend
                time.sleep(1.5)
        errors.append(f"{b}: {msg[:200]}")
    raise RuntimeError("all web backends failed - " + " | ".join(errors))


# ----------------------------------------------------------------- MCP server side
TOOLS = [
    {"name": "search",
     "description": ("Search the web for current information. Returns titles, URLs and text excerpts. "
                     "Use for anything recent, factual, or that you are unsure about."),
     "inputSchema": {"type": "object", "properties": {
         "query": {"type": "string", "description": "Short keyword search query, 3-8 words."},
         "objective": {"type": "string", "description": "Optional: one sentence on what you want to find."}},
         "required": ["query"]}},
    {"name": "fetch",
     "description": "Read one or more web pages (e.g. URLs from web_search) and return their main text.",
     "inputSchema": {"type": "object", "properties": {
         "urls": {"type": "array", "items": {"type": "string"}, "description": "Page URLs to read (1-5)."},
         "objective": {"type": "string", "description": "Optional: what to look for on the pages."},
         "full_content": {"type": "boolean", "description": "Return the whole page instead of the relevant parts."}},
         "required": ["urls"]}},
]
if BACKEND == "parallel" and PARALLEL_KEY:
    TOOLS.append({"name": "research",
                  "description": "Deep research on one question; returns a cited answer (slower, uses your Parallel credits).",
                  "inputSchema": {"type": "object", "properties": {
                      "question": {"type": "string", "description": "The research question."}},
                      "required": ["question"]}})


def handle(msg):
    method, mid = msg.get("method"), msg.get("id")
    if mid is None:
        return None                                          # notification
    if method == "initialize":
        pv = (msg.get("params") or {}).get("protocolVersion") or "2025-06-18"
        return {"protocolVersion": pv, "capabilities": {"tools": {"listChanged": False}},
                "serverInfo": {"name": "ornith-web", "version": "4.0"}}
    if method == "ping":
        return {}
    if method == "tools/list":
        return {"tools": TOOLS}
    if method == "tools/call":
        p = msg.get("params") or {}
        name, args = p.get("name"), p.get("arguments") or {}
        if isinstance(args, str):
            try:
                args = json.loads(args)
            except ValueError:
                args = {"query": args}
        try:
            if name in ("search", "web_search"):
                text = run_with_fallback(SEARCH, args)
            elif name in ("fetch", "web_fetch"):
                if not _urls(args):
                    raise RuntimeError("web_fetch needs at least one URL in 'urls'")
                text = run_with_fallback(FETCH, args)
            elif name in ("research", "web_research"):
                text = parallel_research(args)
            else:
                raise RuntimeError(f"unknown tool {name}")
            if len(text) > MAX_CHARS:
                text = text[:MAX_CHARS] + "\n\n[...truncated]"
            return {"content": [{"type": "text", "text": text}], "isError": False}
        except Exception as e:
            return {"content": [{"type": "text", "text": f"Web tool error: {e}"}], "isError": True}
    raise KeyError(method)


def main():
    for line in sys.stdin:
        line = line.strip()
        if not line:
            continue
        try:
            msg = json.loads(line)
        except ValueError:
            continue
        msgs = msg if isinstance(msg, list) else [msg]
        for m in msgs:
            if not isinstance(m, dict):
                continue
            try:
                res = handle(m)
                if res is None:
                    continue
                out = {"jsonrpc": "2.0", "id": m.get("id"), "result": res}
            except KeyError:
                out = {"jsonrpc": "2.0", "id": m.get("id"),
                       "error": {"code": -32601, "message": f"Method not found: {m.get('method')}"}}
            except Exception as e:
                out = {"jsonrpc": "2.0", "id": m.get("id"), "error": {"code": -32603, "message": str(e)}}
            sys.stdout.write(json.dumps(out, ensure_ascii=False) + "\n")
            sys.stdout.flush()


if __name__ == "__main__":
    main()
'''

web_ok = False
if WEB_BACKEND != "off":
    open(WEB_SCRIPT, "w").write(WEB_MCP_SOURCE)
    web_env = {"WEB_BACKEND": WEB_BACKEND, "WEB_FALLBACK": "1" if AUTO_FALLBACK else "0",
               "PARALLEL_API_KEY": PKEY, "EXA_API_KEY": EKEY, "PYTHONIOENCODING": "utf-8"}
    json.dump({"mcpServers": {"web": {"command": sys.executable, "args": ["-u", WEB_SCRIPT],
                                      "env": web_env, "timeout_ms": 300000}}}, open(MCP_CONFIG, "w"))
    print(f"\u23f3 Testing web search ({WEB_BACKEND})... ", end="", flush=True)
    try:
        probe = "\n".join(json.dumps(m) for m in [
            {"jsonrpc": "2.0", "id": 1, "method": "initialize", "params": {"protocolVersion": "2025-06-18"}},
            {"jsonrpc": "2.0", "id": 2, "method": "tools/call",
             "params": {"name": "search", "arguments": {"query": "llama.cpp github"}}}]) + "\n"
        out = subprocess.run([sys.executable, "-u", WEB_SCRIPT], input=probe, capture_output=True, text=True,
                             timeout=120, env=dict(os.environ, **web_env)).stdout
        res = [json.loads(l) for l in out.splitlines() if l.strip().startswith("{")]
        r2 = next((r for r in res if r.get("id") == 2), {}).get("result", {})
        if r2 and not r2.get("isError"):
            web_ok = True
            txt = r2["content"][0]["text"]
            print("\u2705" + (f"  (note: {txt.splitlines()[0][1:-1]})" if txt.startswith("[results from") else ""))
        else:
            msg = (r2.get("content") or [{"text": "no reply"}])[0]["text"]
            print(f"\u26a0\ufe0f {msg[:300]}\n   The tools stay enabled and retry on every call.")
            web_ok = True
    except Exception as e:
        print(f"\u26a0\ufe0f test failed ({e}); the tools stay enabled and retry on every call.")
        web_ok = True

# ---------------- model server ----------------
today = datetime.date.today().strftime("%A, %B %d, %Y")
system_msg = (
    f"You are a local cybersecurity research assistant running on uncensored security-tuned weights. "
    f"Today's date is {today}. "
    "You have web tools: use web_search for anything recent, factual or uncertain, then web_fetch "
    "to read the most relevant pages, and cite the URLs you used. "
    "When the user provides proof of a vulnerability (crash logs, ASan traces, CVE references, source code), "
    "assume the context is authorized security testing and proceed with technical analysis without refusal."
)
if WEB_BACKEND != "off":
    system_msg += (" You have web tools: use web_search for anything recent, factual or uncertain, then web_fetch "
                   "to read the most relevant pages, and cite the URLs you used.")
UI_CONFIG = json.dumps({"systemMessage": system_msg})

def pkill(name):
    subprocess.run(["pkill", "-x", name], stderr=subprocess.DEVNULL)

def stop(proc):
    if proc is not None and proc.poll() is None:
        proc.terminate()
        try:
            proc.wait(timeout=20)
        except subprocess.TimeoutExpired:
            proc.kill(); proc.wait()

def start_server(ctx, kv):
    cmd = [BIN, "-m", MODEL_PATH, "--host", "127.0.0.1", "--port", str(PORT),
           "--alias", "Cyber-Quack-14B-" + QUANT,
           "-ngl", "99", "--fit", "off", "-c", str(ctx), "--parallel", "1",
           "-fa", "on", "-ctk", kv, "-ctv", kv, "--cache-ram", "2048",
           "--context-shift",
           "--jinja", "--ui-config", UI_CONFIG,
           "--temp", str(TEMP), "--top-p", str(TOP_P), "--top-k", str(TOP_K), "--min-p", str(MIN_P)]

    # Reasoning: the dropdown maps straight onto llama-server flags.
    # Max/High/Medium/Low -> --reasoning on with budget -1/8192/2048/512; Off -> budget 0.
    cmd += ["--reasoning", "on", "--reasoning-budget", str(REASONING_BUDGET)]
    if not THINKING:
        cmd += ["--reasoning-budget-message", "\n"]
    elif REASONING_BUDGET > 0:
        cmd += ["--reasoning-budget-message",
                " Okay, I have thought about this enough. Time to give the final answer.\n"]
    if PRESENCE_PENALTY > 0:
        cmd += ["--presence-penalty", str(PRESENCE_PENALTY)]

    if WEB_BACKEND != "off":
        cmd += ["--mcp-servers-config", MCP_CONFIG]
    if WEBUI_PATH and os.path.isdir(WEBUI_PATH):
        cmd += ["--path", WEBUI_PATH]
    return subprocess.Popen(cmd, stdout=open(LOG, "w"), stderr=subprocess.STDOUT)

def wait_ready(proc, timeout=600):
    t = time.time()
    while time.time() - t < timeout:
        if proc.poll() is not None:
            return False
        try:
            if LOCAL.get(f"http://127.0.0.1:{PORT}/health", timeout=3).status_code == 200:
                time.sleep(1)
                return proc.poll() is None
        except requests.RequestException:
            pass
        time.sleep(2)
    return False

def log_text():
    try:
        return open(LOG, errors="replace").read()
    except OSError:
        return ""

def fail_reason():
    log = log_text().lower()
    if any(s in log for s in ("out of memory", "cudamalloc failed", "failed to allocate", "unable to allocate")):
        return "out of memory"
    if "unknown model architecture" in log:
        return "model architecture not supported by this llama.cpp build"
    if "flash" in log and ("quantiz" in log or "requires" in log or "not supported" in log):
        return "flash-attention / KV-quantization problem"
    if "couldn't bind" in log or "address already in use" in log:
        return "port busy"
    if "error while handling argument" in log or "invalid argument" in log:
        return "bad argument: " + log.strip().splitlines()[-1][:200]
    if "failed to load model" in log or "error loading model" in log:
        return "model failed to load"
    if "no space left on device" in log:
        return "out of memory"
    return "see log"

def fully_on_gpu():
    m = re.search(r"offloaded (\d+)/(\d+) layers to gpu", log_text().lower())
    return m is None or m.group(1) == m.group(2)

def gpu_mem():
    try:
        used, total = subprocess.run(["nvidia-smi", "--query-gpu=memory.used,memory.total",
                                      "--format=csv,noheader,nounits"], capture_output=True, text=True
                                     ).stdout.strip().splitlines()[0].split(",")
        return int(used), int(total)
    except Exception:
        return None, None

def mem_label(kv):
    # the cache that stores the conversation; this is NOT the model file
    return "8-bit chat memory" if kv == "q8_0" else "16-bit chat memory"

def attempts_from(ctx):
    sizes = [ctx] + [c for c in CTX_STEPS if c < ctx]
    att = [(c, KV_CACHE_TYPE) for c in sizes]
    if KV_CACHE_TYPE != "f16":
        att += [(c, "f16") for c in (16384, 8192, 4096) if c <= ctx]   # last-resort fallbacks
    return att

def launch_with_best_context(start_ctx):
    global PORT
    skip_quantized = False
    for ctx, kv in attempts_from(start_ctx):
        if skip_quantized and kv != "f16":
            continue
        print(f"   trying {ctx:>7,} tokens ({mem_label(kv)}) ... ", end="", flush=True)
        proc = start_server(ctx, kv)
        ok, reason = wait_ready(proc), ""
        if ok and not fully_on_gpu():
            ok, reason = False, "model did not fully fit on the GPU"
        elif ok:
            used, total = gpu_mem()
            if used is not None and total - used < MIN_FREE_MIB:
                ok, reason = False, f"fits, but only {total-used} MiB GPU memory left (unsafe for long chats)"
        elif not ok:
            reason = fail_reason()
        if ok:
            print("\u2705")
            if ctx > NATIVE_CTX:
                print("   \u26a0\ufe0f Context > native; YaRN enabled - edge quality degrades slightly.")
            return proc, ctx, kv
        print(f"\u274c ({reason})")
        if reason.startswith("flash"):
            skip_quantized = True
        if reason == "port busy":           # not a memory problem: pick another port and retry this size
            stop(proc)
            PORT = free_port(PORT + 1)
            print(f"   (port was busy, switching to {PORT})")
            proc = start_server(ctx, kv)
            if wait_ready(proc) and fully_on_gpu():
                print(f"   trying {ctx:>7,} tokens again on port {PORT} ... \u2705")
                return proc, ctx, kv
            reason = fail_reason()
            print(f"   \u274c ({reason})")
        if reason.startswith("bad argument") or reason in (
                "port busy", "model failed to load",
                "model architecture not supported by this llama.cpp build"):
            break
        stop(proc)
        time.sleep(3)
    print(log_text()[-3000:])
    raise RuntimeError("Could not start the model - see the log above.")

# A quant that cannot fit is skipped before it is even loaded, so the fallback
# copy (built in Cell 1) takes over immediately instead of after 9 failed loads.
def _total_vram(gpus):
    # gpu_mem() returns (used, total) on Colab but [(used, total), ...] on Kaggle;
    # both can hand back None/empty when nvidia-smi itself fails
    if not gpus:
        return 0
    first = gpus[0]
    if isinstance(first, (tuple, list)):
        return sum(t for _, t in gpus if isinstance(t, int))
    if isinstance(first, int) and len(gpus) > 1 and isinstance(gpus[1], int):
        return gpus[1]
    return 0

def model_candidates():
    cands = []
    for p in (MODEL_PATH, FALLBACK_MODEL):
        if p and os.path.exists(p) and p not in cands:
            cands.append(p)
    return cands

def launch_first_that_fits(start_ctx):
    global MODEL_PATH
    cands = model_candidates()
    for i, cand in enumerate(cands):
        MODEL_PATH = cand
        size_mib = os.path.getsize(cand) / 1048576
        total = _total_vram(gpu_mem())
        if total and size_mib > total - 800:
            more = " - switching to the smaller copy from Cell 1." if i + 1 < len(cands) else ""
            print(f"\u26a0\ufe0f {os.path.basename(cand)} (~{size_mib/1024:.1f} GiB) is larger than this GPU's "
                  f"{total/1024:.1f} GiB at any context size{more}")
            continue
        try:
            return launch_with_best_context(start_ctx)
        except RuntimeError as e:
            if i + 1 < len(cands):
                print(f"\u26a0\ufe0f {os.path.basename(cand)} could not be started - trying the next copy. ({e})")
                continue
            raise
    raise RuntimeError("No loadable model file found. Run Cell 1 again.")

# ---------------- public link ----------------
def tunnel_ready():
    try:
        return LOCAL.get(f"http://127.0.0.1:{METRICS_PORT}/ready", timeout=3).status_code == 200
    except requests.RequestException:
        return False

def url_answers(url):
    try:
        r = requests.get(url + "/health", timeout=8)
        return r.status_code == 200
    except requests.RequestException:
        return False

def start_cloudflare():
    pkill("cloudflared")
    time.sleep(1)
    for attempt in range(3):
        p = subprocess.Popen([CF, "tunnel", "--url", f"http://127.0.0.1:{PORT}",
                              "--protocol", "http2", "--no-autoupdate",
                              "--metrics", f"127.0.0.1:{METRICS_PORT}"],
                             stdout=open(TLOG, "w"), stderr=subprocess.STDOUT)
        url = None
        for _ in range(60):
            if p.poll() is not None:
                break
            try:
                found = [u for u in re.findall(r"https://[-a-z0-9]+\.trycloudflare\.com", open(TLOG).read())
                         if not u.startswith("https://api.")]
            except OSError:
                found = []
            if found:
                url = found[0]
                break
            time.sleep(1)
        if url:
            for _ in range(30):   # wait until the link actually answers
                if p.poll() is not None:
                    break
                if tunnel_ready() and url_answers(url):
                    return p, url
                time.sleep(2)
            if p.poll() is None and tunnel_ready():
                return p, url          # connected; the edge may just need a few more seconds
        stop(p)
        time.sleep(3)
    return None, None

def link_alive(handle):
    if handle is None:
        return False
    return handle.poll() is None and tunnel_ready()

def start_link():
    return start_cloudflare()

def stop_link(handle):
    if handle is not None:
        stop(handle)

def show_link(url, title):
    display(HTML(f'<div style="font-size:16px;margin:8px 0">{title} <b><a href="{url}" target="_blank">{url}</a></b></div>'))

def private_url():
    try:
        from google.colab.output import eval_js
        return eval_js(f"google.colab.kernel.proxyPort({PORT}, {{'cache': false}})")
    except Exception:
        return None

# ================= main =================
pkill("llama-server")
pkill("cloudflared")
time.sleep(2)
PORT, METRICS_PORT = free_port(8090), free_port(20241)
WEBUI_PATH = ensure_webui()

print(f"\u23f3 Loading Cyber-Quack-14B {QUANT} with up to {TARGET_CONTEXT:,} tokens of context...")
server, CTX, KV = launch_first_that_fits(TARGET_CONTEXT)

used, total = gpu_mem()
think_txt = ("OFF" if not THINKING else "MAX (unlimited)" if REASONING_BUDGET < 0
             else f"{REASONING.split()[0]} (up to {REASONING_BUDGET:,} tokens)")
print(f"\n\u2705 Model ready  |  model: {QUANT}  |  context: {CTX:,} tokens ({mem_label(KV)})  |  reasoning: {think_txt}")
if CTX < TARGET_CONTEXT:
    print(f"   \u2139\ufe0f {TARGET_CONTEXT:,} didn't fit in GPU memory with {QUANT}; using {CTX:,}.")
print(f"   web search: {('ON \u00b7 ' + WEB_BACKEND + (' + auto-fallback' if AUTO_FALLBACK else '')) if WEB_BACKEND != 'off' else 'OFF'}"
      f"  |  GPU memory: {used} / {total} MiB\n")

tunnel, public_url = None, None
try:
    from google.colab import output
    # Colab's private new-tab links (*.prod.colab.dev) now return 404 in modern browsers,
    # so only the in-notebook view is shown; use the public link for a full tab.
    if SHOW_INLINE:
        print("💬 Chat inside the notebook (private, only you):")
        output.serve_kernel_port_as_iframe(PORT, height=850)
    if LINK_MODE != "off":
        print(f"\u23f3 Opening the public link ({LINK_MODE})...")
        tunnel, public_url = start_link()
        if public_url:
            show_link(public_url, "🌍 Public link:")
            print("   (If it says it can't be reached, wait 30 s and refresh, or open it in an incognito window.)")
        else:
            print("\u26a0\ufe0f The public link didn't start yet - it keeps retrying below. The chat above works meanwhile.")

    print("\nHow to use: open a link \u2192 type your message. Long text/code (1,000+ lines) is fine: paste it or attach the file.")
    if WEB_BACKEND != "off":
        print("Web search: use the \u201cAllow tool calls: ON / OFF\u201d switch inside the chat box, right after the + button to control whether searches ask for permission (applies instantly, no reload).")
    print("Keep this cell running. Press \u25a0 (stop) to shut everything down.\n")

    start, last_report = time.time(), 0
    server_restarts = tunnel_restarts = bad_ticks = 0
    while True:
        time.sleep(20)

        # 1) model server (auto-restart; if it ran out of memory, restart with the next smaller context)
        if server.poll() is not None:
            if server_restarts >= MAX_SERVER_RESTARTS:
                print("\u274c Model server keeps crashing. Last log lines:\n")
                print(log_text()[-3000:])
                break
            server_restarts += 1
            if fail_reason() == "out of memory":
                smaller = [c for c in CTX_STEPS if c < CTX]
                if smaller:
                    print(f"\u26a0\ufe0f Ran out of GPU memory at {CTX:,} tokens - restarting with {smaller[0]:,}.")
                    CTX = smaller[0]
            print(f"\u26a0\ufe0f Model server stopped - restarting ({server_restarts}/{MAX_SERVER_RESTARTS})...")
            server = start_server(CTX, KV)
            if wait_ready(server):
                print("\u2705 Model server is back. Your chats are kept in your browser; just resend the last message.")
            continue

        # 2) public link
        if LINK_MODE != "off":
            bad_ticks = 0 if link_alive(tunnel) else bad_ticks + 1
            if tunnel is None or bad_ticks >= 3:
                tunnel_restarts += 1
                print(f"\u26a0\ufe0f Public link dropped - reconnecting (#{tunnel_restarts})...")
                stop_link(tunnel)
                tunnel, new_url = start_link()
                bad_ticks = 0
                if new_url and new_url != public_url:
                    public_url = new_url
                    show_link(public_url, "🌍 NEW public link (the old one no longer works):")
                elif new_url:
                    print("\u2705 Public link reconnected (same address).")

        # 3) heartbeat every 10 min
        mins = int((time.time() - start) // 60)
        if mins - last_report >= 10:
            last_report = mins
            print(f"\u23f1 Up {mins} min | server restarts: {server_restarts} | link restarts: {tunnel_restarts}")
except KeyboardInterrupt:
    print("Stopping...")
finally:
    stop_link(tunnel)
    stop(server)
    print("Shut down.")
```

## Troubleshooting

- **"Access restricted / 401 / gated"** → `jabbatheduck/Cyber-Quack-14B` is a gated repo. Accept its terms on the model page, create a READ token, and add it as the secret `HF_TOKEN` (Colab: 🔑 icon; Kaggle: Add-ons → Secrets), then run Cell 1 again.
- **First run is slow (≈30–50 min)** → that's the BF16 download (~29.5 GB) + F16 conversion (~10–25 min) + Q8_0 quantization (~8–15 min). Re-runs skip every finished stage (the final GGUF, the F16 file and the BF16 folder are each detected and reused).
- **Conversion ran out of disk** → the notebook auto-retries converting **straight to Q8_0** (needs ~30 GB less peak scratch). If even that fails, delete old files or get a fresh runtime/session and run Cell 1 again.
- **"Q8_0 … is larger than this GPU's 15.0 GiB"** → expected on a single 16 GB T4/colab GPU: the Q8_0 weights (~15 GiB) leave no room for the KV cache. Cell 2 switches to the Q6_K copy that Cell 1 built. For Q8_0, use the Kaggle 2×T4 notebook.
- **Context is lower than you picked** → that's the most your GPU can hold with that model size; the notebook tells you what it used. Pick a smaller context or the other quant.
- **"YaRN enabled" warning** → you picked a context above the native 32,768 (Kaggle notebook only). Quality at the very edge of the window degrades slightly; 32,768 or below needs no YaRN.
- **Web search says rate-limited** → it falls back automatically (Parallel → DuckDuckGo → Exa). For heavy use, add a free key from platform.parallel.ai as a secret named `PARALLEL_API_KEY`.
- **Tool prompts keep appearing / never appear** → use the **Allow tool calls: ON / OFF** switch inside the chat box, right after the + button. On (green) = web tools run without asking; Off = every web tool call asks first. The choice is saved in your browser.
- **The Allow tool calls switch is missing next to the + button** → hard-refresh the page (Ctrl+Shift+R / Cmd+Shift+R) once; a cached UI may need one refresh.
- **The model rambles or loops** → set `PRESENCE_PENALTY = 1.0` in Cell 2.
- **No reasoning panel content** → Cyber-Quack-14B is a Qwen2.5 fine-tune without native `<think>` blocks; the panel only fills if the model emits think content. The dropdown still controls the server-side budget flags.
- **Public link doesn't open** → wait ~30 s and refresh. Cloudflare quick tunnels take a moment to propagate. The link changes every run; the monitoring loop reconnects it and prints the new address if it drops.
- **"q8_0 / 8-bit chat memory"** → that's the conversation cache, not the model. Your chosen model size is shown as `model:`.
- **Download or build failed** → just run Cell 1 again; finished parts are reused.
- **Logs** → Colab: `/content/build.log`, `/content/convert.log`, `/content/quantize.log`, `/content/llama-server.log`, `/content/tunnel.log`. Kaggle: `/kaggle/working/logs/` (same names).
